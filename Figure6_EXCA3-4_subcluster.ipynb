```r
scRNA.sub = readRDS("EX_CA3-4.subcluster.rds")
```

```r
print("1")
scRNA.sub = SCTransform(scRNA.sub)
scRNA.sub = NormalizeData(scRNA.sub)   
scRNA.sub = FindVariableFeatures(scRNA.sub, selection.method = "vst")
scRNA.sub = ScaleData(scRNA.sub)
scRNA.sub <- RunPCA(scRNA.sub, features = VariableFeatures(scRNA.sub))
```

```r
head(scRNA.sub)
```

```r
scRNA.sub=RunHarmony(scRNA.sub,group.by.vars="sample", max.iter.harmony = 20) %>% 
                FindNeighbors(reduction = "harmony", dims = 1:20)%>% 
                FindClusters(resolution =c(0.3,0.5,0.8,1))
scRNA.sub = RunUMAP(scRNA.sub, reduction = "harmony", dims = 1:30) 
```
